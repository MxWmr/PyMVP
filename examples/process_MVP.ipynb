```python
import PyMVP as mvp

path = "WHIRLS_mvp/SECTIONS_9_12/"
savepath = "MVP_pictures/"
savepath_ncdf= "MVP_Data_processed/transect_NCDF/"
mvpa = mvp.Analyzer()
offset_temp=0
offset_cond = -0.015
offset_oxy =  -13.75
offset_oxy =  -11.82
mvpa.load_mvp_data(path,subdirs=True,delp=[],offset_temp=offset_temp,offset_cond=offset_cond,offset_oxy=offset_oxy)


mvpa.print_profile_metadata()



mvpa.load_gps_from_ncdf('WHIRLS_nav/')



mvpa.mvp_correction()
mvpa.correct_oxygen_with_slope(0.00361)


path_cal = "WHIRLS_mvp/SECTIONS_9_12/SNA2338C.cal"


mvpa.process_NO3(path_cal)

mvpa.interpolate_CTD_and_MVPcorrected(10000)

mvpa.compute_dist()
```
